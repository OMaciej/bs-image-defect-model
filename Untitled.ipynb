```python
import cv2
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('data/partialdata.csv')
```

```python
for scene, name, edge, gauss, blur, unreal in df.itertuples(index=False):
    if edge < 0.01 and gauss < 0.01 and unreal < 0.01 and blur < 0.01:
        break
```

```python
import os
```

```python
img = cv2.imread(os.path.join('data/images', scene, name))
img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
plt.imshow(img_rgb, vmin=0, vmax=255)
```

```python
plt.imshow(img_gray, cmap='gray')
```

```python
img_edges = cv2.Canny(img_gray, 100, 200)
plt.imshow(img_edges, cmap='gray')
```

```python
print(edge)
print(gauss)
print(blur)
print(unreal)
```

```python
grad_x = cv2.Sobel(img_gray, cv2.CV_32F, 1, 0, ksize=3)
grad_y = cv2.Sobel(img_gray, cv2.CV_32F, 0, 1, ksize=3)
```

```python
abs_grad_x = cv2.convertScaleAbs(grad_x)
abs_grad_y = cv2.convertScaleAbs(grad_y)
    
    
grad = cv2.addWeighted(abs_grad_x, 0.5, abs_grad_y, 0.5, 0)
```

```python
plt.imshow(grad, cmap='gray')
```

```python
lap = cv2.Laplacian(img_gray, cv2.CV_32F)
```

```python
plt.imshow(lap, cmap='gray')
```
